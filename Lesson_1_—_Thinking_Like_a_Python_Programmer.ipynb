```python
# ==========================================
# LESSON 1: PYTHON FOUNDATIONS
# ==========================================


# ------------------------------------------
# 1. VARIABLES
# ------------------------------------------
# A variable stores information.
# Think of it like a labeled container.

name = "Mariah"
age = 25
gpa = 4.0
mcc_student = True


# ------------------------------------------
# 2. BASIC DATA TYPES
# ------------------------------------------

# int = whole numbers
age = 25

# float = numbers with decimals
gpa = 4.0

# str = text
name = "Mariah"

# bool = True or False
mcc_student = True


# ------------------------------------------
# 3. CHECKING DATA TYPES
# ------------------------------------------
# type() tells you what type of data a variable contains.

print(type(age))           # int
print(type(gpa))           # float
print(type(name))          # str
print(type(mcc_student))   # bool


# ------------------------------------------
# 4. PRINTING INFORMATION
# ------------------------------------------
# print() displays information in the output.

print(name)
print(age)
print(gpa)
print(mcc_student)


# ------------------------------------------
# 5. BASIC MATH
# ------------------------------------------

# Addition
total = 100 + 50

# Subtraction
difference = 100 - 50

# Multiplication
product = 100 * 50

# Division
average = 100 / 50

# Exponent
squared = 5 ** 2

print(total)
print(difference)
print(product)
print(average)
print(squared)


# ------------------------------------------
# 6. CALCULATING AN AVERAGE
# ------------------------------------------
# Step 1: Create variables for the scores.

quiz1 = 85
quiz2 = 92
quiz3 = 78

# Step 2: Calculate the average.

average_score = (quiz1 + quiz2 + quiz3) / 3

# Step 3: Inspect the result.

print(average_score)


# ------------------------------------------
# 7. THE CREATE → CALCULATE → INSPECT
#    MENTAL MODEL
# ------------------------------------------

# CREATE
sales = 100000
expenses = 45000

# CALCULATE
profit = sales - expenses

# INSPECT
print(profit)


# ------------------------------------------
# 8. IMPORTANT: NUMBERS AND COMMAS
# ------------------------------------------
# DO NOT use commas when entering numbers.

salary = 52000       # Correct
# salary = 52,000   # Incorrect


# If you want to represent fifty-two thousand,
# write 52000, NOT 52,000.


# ------------------------------------------
# 9. VARIABLE NAMING
# ------------------------------------------
# Use descriptive variable names.
# Python convention is lowercase with underscores
# between words (snake_case).

employee_1 = 52000
employee_2 = 61000
employee_3 = 73000

total_payroll = employee_1 + employee_2 + employee_3

average_salary = total_payroll / 3

print(total_payroll)
print(average_salary)


# ------------------------------------------
# 10. AVOID VAGUE VARIABLE NAMES
# ------------------------------------------
# This works, but doesn't tell you what the
# information represents.

x = 52000
y = 61000
z = 73000


# This is better because the variable names
# explain what the values represent.

employee_1 = 52000
employee_2 = 61000
employee_3 = 73000


# ------------------------------------------
# 11. WHEN YOU SEE A BLANK CODE CELL
# ------------------------------------------
# Ask yourself:
#
# 1. What information do I have?
# 2. What do I need to find?
# 3. What calculation or action needs to happen?
# 4. How can I represent that in Python?
#
# Then use:
#
# CREATE → CALCULATE → INSPECT


# ------------------------------------------
# 12. LESSON 1 KEY TAKEAWAYS
# ------------------------------------------

# Variables store information.

# Python has different data types:
# int, float, str, and bool.

# print() displays information.

# type() tells you the data type.

# Python can perform basic math using:
# +   addition
# -   subtraction
# *   multiplication
# /   division
# **  exponent

# Do NOT use commas when entering numbers.
# 52000 is correct.
# 52,000 is NOT the same thing in Python.

# Use descriptive variable names.
# snake_case is the standard style.

# Most importantly:
# Break problems into small steps.
#
# CREATE → CALCULATE → INSPECT
```

```python
# remember the pattern: variable[index]
# Negative indexes count backward from the end

name = "Mariah"
print(name[0], name[2], name[5])

city = "Phoenix"
print(city[-1]) # print last character
```

```python
# String indexing breakdown for "Phoenix":
#
# Positive indices:
#   P    h    o    e    n    i    x
#   0    1    2    3    4    5    6
#
# Negative indices:
#  -7   -6   -5   -4   -3   -2   -1
```

```python
word = "Python"
print(word[-1],word[-2])
```

```python
# word[start:stop]
# Start is included. Stop is excluded.
# String slicing breakdown for "Python":
#
# Start at index 0  → 'P'
# Include index 1   → 'y'
# Include index 2   → 't'
# Stop before index 3
#
# Result: "Pyt"

# Print: Pyt
print(word[0:3])
print(word[2:6])
print(word[0:])
print(word[2:])   # from index 2 to the end
print(word[:4])    # from the beginning up to index 4
print(word[:])     # entire string
print(word[0],word[1],word[2])
```

```python
email = "  Mariah.Maloof@GMAIL.COM  "

# 1. Strips outer spaces and converts to lowercase
new_email = email.strip().lower()  # "mariah.maloof@gmail.com"

# 2. Splits on "@" and picks index 1 (the domain)
print(new_email.split("@")[1])  # Output: gmail.com

# 3. Searches for "#"
email.find("#")  # Returns -1 because "#" is not in the string
```
